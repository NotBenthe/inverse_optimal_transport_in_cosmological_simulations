```python
import numpy as np

epsilon = 0.1

C = np.array([[1, 9], 
              [1, 1]])
K = np.exp(-C / epsilon)

mu = np.array([1/3, 2/3]) 
nu = np.array([2/3, 1/3])
v = np.array([0.5, 0.5]) # v^0

print("Kernel Matrix K:")
print(K)
print("-" * 30)

def sinkhorn_iteration(K, mu, nu, v_old):
    # u = mu / (K @ v)
    u_new = mu / np.dot(K, v_old)
    # v = nu / (K.T @ u)
    v_new = nu / np.dot(K.T, u_new)
    return u_new, v_new

u1, v1 = sinkhorn_iteration(K, mu, nu, v)
print(f"Iteration 1:\nu1: {u1}\nv1: {v1}\n")

u2, v2 = sinkhorn_iteration(K, mu, nu, v1)
print(f"Iteration 2:\nu2: {u2}\nv2: {v2}\n")

diff_u = np.linalg.norm(u2 - u1)
diff_v = np.linalg.norm(v2 - v1)

print(f"Frobenius Norm ||u2 - u1||: {diff_u:.2f}")
print(f"Frobenius Norm ||v2 - v1||: {diff_v:.2f}")

# running for 20 iterations
u_curr, v_curr = u2, v2
for i in range(3, 21):
    u_next, v_next = sinkhorn_iteration(K, mu, nu, v_curr)
    if i == 20:
        dist_u = np.linalg.norm(u_next - u_curr)
        dist_v = np.linalg.norm(v_next - v_curr)
        print("-" * 30)
        print(f"At 20 iterations:")
        print(f"Difference in u: {dist_u:.2e}")
        print(f"Difference in v: {dist_v:.2e}")
    u_curr, v_curr = u_next, v_next

pi_hat = u_curr[:, None] * K * v_curr[None, :]
print(pi_hat)
```
